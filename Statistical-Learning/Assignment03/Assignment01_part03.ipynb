It contains a number of variables for **506 different house in Boston state**. The
variables are:

- crim: per capita crime rate by town
- zn: proportion of residential land zoned for lots over 25,000 sq.ft.
- indus: proportion of non-retail business acres per town.
- chas: Nearby Charles River or not
- nox: nitric oxides concentration
- rm: average number of rooms per dwelling
- age: proportion of owner-occupied units built prior to 1940
- dis: weighted distances to five Boston employment centres
- rad: index of accessibility to radial highways
- tax: full-value property-tax rate
- ptratio: pupil-teacher ratio by town
- lstat: % lower status of the population
- medv: Median value of owner-occupied homes

```python
# (a) load the data

import pandas as pd
boston = pd.read_csv('/content/Boston.csv')
```

```python
"""
  (b) How many rows are in this data set?
  How many columns?
"""

print(boston.shape)
boston = boston.rename ({'Unnamed: 0': 'House'},axis =1)
print(boston.head())
```

```python
# (c) Make some pairwise scatterplots of the predictors (columns) in this data set.
import matplotlib.pyplot as plt

cols = boston.select_dtypes(include=['number']).columns
n = len(cols)
fig, axes = plt.subplots(nrows=n, ncols=n, figsize=(20, 20))

for i in range(n):
  for j in range(n):
    col1 = cols[i]
    col2 = cols[j]
    ax = axes[i, j]
    if col1 != col2:
      ax.scatter(boston[col2], boston[col1], alpha=0.5, s=5)
    if i == n - 1:
      ax.set_xlabel(col2, fontsize=8)
    if j == 0:
      ax.set_ylabel(col1, fontsize=8)
    ax.set_xticks([])
    ax.set_yticks([])
plt.tight_layout()
plt.subplots_adjust(wspace=0.1, hspace=0.1)
plt.show()
```
