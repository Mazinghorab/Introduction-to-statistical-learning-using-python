It contains a number of variables for **397 different Autos company**. The
variables are:

- mpg: How far your car can travel for every gallon
- cylinders: The fuel is combusted and power is generated
- displacement: Indicator of the power of the engine
- horsepower: The output of engines or motors
- weight: Weight of a it's cars
- acceleration: Speed of it's car Models
- year: Company founding year
- origin: Classes of each car
- name: Name of the car Company

```python
# read the data
import pandas as pd
auto = pd.read_csv("/content/Auto.csv", na_values=['?'])
```

```python
# let's understand what is this data and try to clean it
print(auto.head())
print(auto.info())
print(auto.shape)
auto = auto.dropna()
auto = auto.set_index('name')
print("After: \n")
print(auto.head())
print(auto.info())
print(auto.shape)
```

```python
# (a) Which of them is qualitative or quantitative?
print(auto.info())
```

```python
# (b) What is the range of each quantitative predictor?

print(auto.describe())
```

```python
# (b) Another Solution

col = auto.columns
for i in col:
  print("The minimum os the column", i, "is:", min(auto[i]))
  print("The maximum os the column", i, "is:", max(auto[i]))
  print("------")
```

```python
# (c) What is the mean and standard deviation of each quantitative predictor?

print(auto.describe())
```

```python
# Another Solution
import numpy as np

for i in col:
  print("The mean of the column", i, "is:", np.mean(auto[i]))
  print("The standard deviation of the column", i, "is:", np.std(auto[i]))
  print("------")
```
