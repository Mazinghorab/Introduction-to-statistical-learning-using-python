The data is about:

- Advertising : Local advertising budget for company at each location (in thousands of dollars)

- Price : Price company charges for car seats at each site

- ShelveLoc : A factor with levels Bad, Good and Medium indicating the quality of the shelving location for the car seats at each site

- Age : Average age of the local population

- Education : Education level at each location

- Urban : A factor with levels No and Yes to indicate whether the store is in an urban or rural location

- US : A factor with levels No and Yes to indicate whether the store is in the US or not

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

carseats = pd.read_csv('/content/Carseats.csv')
print(carseats.head())
carseats.info()
```

```python
carseats['Urban'] = carseats['Urban'].map({'Yes': 1, 'No': -1})
carseats['US'] = carseats['US'].map({'Yes': 1, 'No': -1})
```

```python
'''
  (a) Fit a multiple regression model to predict Sales using Price,
  Urban, and US.
'''

import statsmodels.api as sm
from ISLP.models import ModelSpec as MS, summarize

predictors = ['Price', 'Urban', 'US']
design = MS(predictors)
X = design.fit_transform(carseats)
y = carseats['Sales']
model1 = sm.OLS(y, X)
results1 = model1.fit()
summarize(results1)
```

```python
'''
  (c) Write out the model in equation form, being careful to handle
  the qualitative variables properly.
'''

b0 = results1.params['intercept']
b1 = results1.params['Price']
b2 = results1.params['Urban']
b3 = results1.params['US']
print(f"Sales = {b0:.3f} + ({b1:.3f} * Price) + ({b2:.3f} * Urban) + ({b3:.3f} * US)")
```

```python
print(carseats['ShelveLoc'].value_counts())
carseats['ShelveLoc'] = carseats['ShelveLoc'].map({'Bad': -1, 'Medium': 0, 'Good': 1})
```

```python
'''
  (e) On the basis of your response to the previous question, fit a
  smaller model that only uses the predictors for which there is
  evidence of association with the outcome.
'''

predictors_reduced = ['Price', 'US']
design_reduced = MS(predictors_reduced)
X_reduced = design_reduced.fit_transform(carseats)
y = carseats['Sales']
model_reduced = sm.OLS(y, X_reduced)
results_reduced = model_reduced.fit()
summarize(results_reduced)
```

```python
'''
  (g) Using the model from (e), obtain 95 % confidence intervals for
  the coefficient(s).
'''

confidence_intervals = results_reduced.conf_int()
confidence_intervals.columns = ['2.5%', '97.5%']
print(confidence_intervals)
```
